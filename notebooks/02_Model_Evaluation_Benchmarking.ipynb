# 02: Model Evaluation & Benchmarking
This notebook evaluates the performance of the anomaly detection models on both synthetic and real-world data.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import os
%matplotlib inline
```

## 1. Model Comparison (Synthetic Data)
Load the metrics generated during the cross-validation phase.

```
metrics_df = pd.read_csv('reports/model_comparison.csv', index_col=0)
metrics_df[['Precision', 'Recall', 'F1-Score']].plot(kind='bar', figsize=(10, 6))
plt.title('Model Performance on Synthetic Data')
plt.ylabel('Score')
plt.xticks(rotation=0)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```

## 2. Real-World Benchmarking
How well did the synthetic-trained model generalize to real data?

```
real_metrics = pd.read_csv('reports/real_data_benchmarking.csv')
plt.figure(figsize=(12, 6))
sns.barplot(data=real_metrics, x='Dataset', y='F1-Score', hue='Model')
plt.title('F1-Score Comparison across Real Datasets')
plt.ylim(0, 1)
plt.show()
```
