# 01: Exploratory Data Analysis & Real-World Integration
This notebook explores the synthetic Microwave Radio data and compares it with the integrated real-world datasets.

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
%matplotlib inline
```

## 1. Synthetic Data Overview
Load the synthetic data generated for the microwave radio scenario.

```
df_syn = pd.read_csv('../data/raw/network_traffic.csv')
df_syn['timestamp'] = pd.to_datetime(df_syn['timestamp'])
print(f'Synthetic Data Shape: {df_syn.shape}')
df_syn.head()
```

### Visualize Synthetic Anomalies
Focus on Throughput and RSL during anomalous periods.

```
plt.figure(figsize=(15, 6))
plt.plot(df_syn['timestamp'][:1000], df_syn['throughput_mbps'][:1000], label='Throughput')
plt.fill_between(df_syn['timestamp'][:1000], 0, 400, where=df_syn['is_anomaly'][:1000]==1, color='red', alpha=0.3, label='Anomaly')
plt.title('Synthetic Throughput with Injected Anomalies')
plt.legend()
plt.show()
```

## 2. Real-World Data Integration
Compare the distributions of real-world datasets with our synthetic data.

```
datasets = {
    'Synthetic': '../data/raw/network_traffic.csv',
    '4G Cellular': '../data/external/4g_cellular_mapped.csv',
    'Network Anomaly': '../data/external/network_anomaly_mapped.csv',
    'SDN Traffic': '../data/external/sdn_traffic_mapped.csv'
}

plt.figure(figsize=(12, 6))
for name, path in datasets.items():
    if os.path.exists(path):
        df = pd.read_csv(path)
        sns.kdeplot(df['throughput_mbps'], label=name, fill=True, alpha=0.3)

plt.title('Throughput Distribution: Synthetic vs Real Data')
plt.xlabel('Throughput (Mbps)')
plt.legend()
plt.show()
```

## 3. Feature Correlation
Check if real-world features show similar correlations to synthetic ones.

```
df_real = pd.read_csv('../data/external/network_anomaly_mapped.csv')
plt.figure(figsize=(10, 8))
sns.heatmap(df_real.corr(), annot=True, cmap='coolwarm')
plt.title('Feature Correlation: Network Anomaly Dataset')
plt.show()
```
